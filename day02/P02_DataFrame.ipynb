# DataFrame的创建

```python
import pandas as pd

# DataFrame可由多个Series组成
mango_series = pd.Series([4,5,6,3,1])
apple_series = pd.Series([5,4,3,0,2])
banana_series = pd.Series([2,3,5,2,7])
# 通过Series创建DataFrame 创建DataFrame对象
df = pd.DataFrame({'mango':mango_series, 'apple':apple_series, 'banana':banana_series})
# type(df)    pandas.DataFrame

# 通过字典创建DataFrame
df = pd.DataFrame({'id':[101,102,103],'name':['张三','李四',"王五"],'age':[20,30,40]})
df

# 通过字典创建DataFrame   在创建对象的同时  指定列的顺序(可以过滤)以及行标签
df = pd.DataFrame({'id':[101,102,103],'name':['张三','李四',"王五"],'age':[20,30,40]},columns=['name','age','id'],index=['aaa','bbb','ccc'])
df
```

# DataFrame的常用属性

```python
df = pd.DataFrame(data={"id": [101, 102, 103], "name": ["张三", "李四", "王五"], "age": [20, 30, 40]},index=["aa", "bb", "cc"])
df
# index	DataFrame的行索引
df.index
# columns	DataFrame的列标签
df.columns
# values	DataFrame的值
df.values   # ndarray类型
# ndim	DataFrame的维度
df.ndim
# shape	DataFrame的形状
df.shape
# size	DataFrame的元素个数
df.size
# dtypes	DataFrame的元素类型
df.dtypes

# T	行列转置
df.T


# 通过中括号 获取列的数据 如果是多列,多列列名需要放到列表中
df[['id','name']]
# loc[]	显式索引，按行列标签索引或切片
# 取出aa和cc数据
df.loc[['aa','cc']]
df.loc[['aa','cc'],['name','age']]
# 取出aa到cc数据
df.loc['aa':'cc','id':'age']

# iloc[]	隐式索引，按行列位置索引或切片
df.iloc[:2,:2]
# at[]	使用行列标签访问单个元素
df.at['bb','name']
# iat[]	使用行列位置访问单个元素
df.iat[1,1]
```

# DataFrame的常用方法

```python
import numpy as np
df = pd.DataFrame(data={"id": [101, 102, 103,104,105,106,101], "name": ["张三", "李四", "王五","赵六","冯七","周八","张三"], "age": [10, 20, 30, 40, None, 60,10]},index=["aa", "bb", "cc", "dd", "ee", "ff","aa"])
df
# head()	查看前n行数据，默认5行
df.head(3)
# tail()	查看后n行数据，默认5行
df.tail(3)
# isin()	元素是否包含在参数集合中
df.isin([103,'赵六',40])
# isna()	元素是否为缺失值
df.isna()
# sum()	求和
type(df['age'])
df[['id','age']].sum()
# mean()	平均值
df['age'].mean()
# min()	最小值
df['age'].min()
# max()	最大值
df['age'].max()
# var()	方差
df['age'].var()
# std()	标准差
df['age'].std()
# median()	中位数
df['age'].median()
# mode()	众数
df.mode()
# quantile()	指定位置的分位数，如quantile(0.5) 类似Series
df['age'].quantile()
# describe()	常见统计信息  include和exclude参数,对应的是类型,不是字段名
df.describe(include=np.number)
# info()	基本信息
# df.info()
# value_counts()	每个元素的个数 一行一行
df.value_counts()
# count()	非空元素的个数
df.count()
# drop_duplicates()	去重
df.drop_duplicates()
# sample()	随机采样
df.sample()
# replace()	用指定值代替原有值
df.replace(np.nan,88)
# equals()	判断两个DataFrame是否相同
df1 = pd.DataFrame(data={"id": [101, 102, 103], "name": ["张三", "李四", "王五"], "age": [10, 20, 30]})
df2 = pd.DataFrame(data={"id": [101, 102, 104], "name": ["张三", "李四", "王五"], "age": [10, 20, 30]})
df1.equals(df2)


# 累积默认按照axis=0
df3 = pd.DataFrame({'A': [2, 5, 3, 7, 4],'B': [1, 6, 2, 8, 3]})
# cummax()	累计最大值
df3.cummax(axis=1)
# cummin()	累计最小值
df3.cummin()
# cumsum()	累计和
df3.cumsum(axis=0)
# cumprod()	累计积
df3.cumprod()

# diff()	一阶差分，对序列中的元素进行差分运算，也就是用当前元素减去前一个元素得到差值，默认情况下，它会计算一阶差分，即相邻元素之间的差值。参数：
# periods：整数，默认为 1。表示要向前或向后移动的周期数，用于计算差值。正数表示向前移动，负数表示向后移动。
# axis：指定计算的轴方向。0 或 'index' 表示按列计算，1 或 'columns' 表示按行计算，默认值为 0。
df3.diff(axis=1)

# sort_index()	按行索引排序
df.sort_index()
# sort_values()	按某列的值排序，可传入列表来按多列排序，并通过ascending参数设置升序或降序
df.sort_values(by=['age','name'],ascending=[False,True])

# nlargest()	返回某列最大的n条数据
df.nlargest(n=3,columns='age')
# nsmallest()	返回某列最小的n条数据
df.nsmallest(n=3,columns='age')

```

# DataFrame的运算

```python
# 1）DataFrame与标量运算  标量与每个元素进行计算
df = pd.DataFrame(
    data={"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"]},
    columns=["name", "age"],
    index=[101, 104, 103, 102],
)
df * 2

# 2）DataFrame与DataFrame运算   根据标签索引进行对位计算，索引没有匹配上的用NaN填充。
df1 = pd.DataFrame(
    data={"age": [10, 20, 30, 40], "name": ["张三", "李四", "王五", "赵六"]},
    columns=["name", "age"],
    index=[101, 102, 103, 104],
)
df2 = pd.DataFrame(
    data={"age": [10, 20, 30, 40], "name": ["张三", "李四", "王五", "田七"]},
    columns=["name", "age"],
    index=[102, 103, 104, 105],
)
df1 + df2
```

# DataFrame的更改操作

```python
# 1.设置行索引
# 创建DataFrame时如果不指定行索引，pandas会自动添加从0开始的索引。
df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
# （1）通过set_index()设置行索引 inplace原地替换
df.set_index('id', inplace=True)
# （2）通过reset_index()重置行索引
df.reset_index(inplace=True)
df
```

```python
# 2.修改行索引名和列名
# （1）通过rename()修改行索引名和列名
df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
df.set_index('id', inplace=True)
df.rename(index={101: "一", 102: "二", 103: "三", 104: "四"}, columns={"age": "年龄", "name": "姓名"}, inplace=True)

# （2）将index和columns重新赋值     df.index/columns本身就是列表
df.index = ["Ⅰ", "Ⅱ", "Ⅲ", "Ⅳ"]
df.columns = ["年齡", "名稱"]
df
```

```python
# 3.添加列 通过 df[“列名”] 添加列
df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
df["phone"] = ["13333333333", "14444444444", "15555555555", "16666666666"]
df
```

```python
# 4.删除列
# （1）通过 df.drop(“列名”, axis=1) 删除，也可是删除行 axis=0
df.drop('phone',axis=1)

# （2）通过 del df[“列名”] 删除
del df['age']
df
```

```python
# 5.插入列
# 通过 insert(loc, column, value) 插入。该方法没有inplace参数，直接在原数据上修改。
df = pd.DataFrame({"age": [20, 30, 40, 10], "name": ["张三", "李四", "王五", "赵六"], "id": [101, 102, 103, 104]})
df.insert(loc = 0, column = 'phone', value = df['age'] * df.index)
df
```

# Pandas日期数据处理初识

```python
# to_datetime介绍
df = pd.DataFrame({"gmv":[100,200,300,400],"trade_date":["2025-01-06","2023-10-31","2023-12-31","2023-01-05"]})
df["ymd"] = pd.to_datetime(df["trade_date"])
# 时间属性访问器对象Series.dt,获取日期数据的年月日星期季度月底年底
df['yy'],df['mm'],df['dd']=df['ymd'].dt.year,df['ymd'].dt.month,df['ymd'].dt.day
df['week']=df['ymd'].dt.day_name()      # 获取星期
df['quarter']=df['ymd'].dt.quarter
df['mend']=df['ymd'].dt.is_month_end
df['yend']=df['ymd'].dt.is_year_end
df
```

```python
# to_datetime介绍
df = pd.DataFrame({"gmv":[100,200,300,400],"trade_date":["2025-01-06","2023-10-31","2023-12-31","2023-01-05"]})
df["ymd"] = pd.to_datetime(df["trade_date"])

# freq：这是 to_period() 方法最重要的参数，用于指定要转换的时间周期频率
# 常见的取值如下：
# 	"D"：按天周期，例如 2024-01-01 会转换为 2024-01-01 这个天的周期。
# 	"W"：按周周期，通常以周日作为一周的结束，比如日期落在某一周内，就会转换为该周的周期表示。
# 	"M"：按月周期，像 2024-05-15 会转换为 2024-05。
# 	"Q"：按季度周期，一年分为四个季度，日期会转换到对应的季度周期，例如 2024Q2 。
# 	"A" 或 "Y"：按年周期，如 2024-07-20 会转换为 2024 。

df['D'] = df['ymd'].dt.to_period('D')
df['W'] = df['ymd'].dt.to_period('W')
df['M'] = df['ymd'].dt.to_period('M')
df['Q'] = df['ymd'].dt.to_period('Q')
df['Y'] = df['ymd'].dt.to_period('Y')
df
```
