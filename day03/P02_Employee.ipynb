# 导包 导入数据

```python
import pandas as pd
df = pd.read_csv(r"D:\AAAjobbing\actfirst\Numpy&Pandas\代码\data\employees.csv")
df.head()
```

```python
# 找出薪资最低、最高的员工 方法很多
df.nsmallest(1,'salary')
df.nlargest(1,'salary')
df.sort_values('salary').iloc[[0,-1]]

# 所有部门id
df['department_id'].drop_duplicates()
df['department_id'].unique()

# 查看每个部门的员工数
df.groupby('department_id')['employee_id'].count().rename('员工数量')

# 绘图
df.groupby('department_id')['employee_id'].count().rename('员工数量').plot(kind='bar')

# 找出平均薪资最高的部门id
df.groupby('department_id')['salary'].mean().nlargest(1)
```
