```python
import pandas as pd

# 读取员工数据
df = pd.read_csv(r"D:\AAAjobbing\actfirst\Numpy&Pandas\data\employees.csv")
df

# 按照部门id进行分组
group_by = df.groupby('department_id')
group_by
```

## 1)查看分组

```python
group_by.groups     # 返回一个字典，字典的键是分组的标签，值是属于该组的所有索引的列表。
group_by.get_group(30)  # 获取分组
```

## 2)按列分组

```python
group_by[['salary']]
```

## 3)按组迭代

```python
for dept_id,group in group_by:
    print(f"当前组为{dept_id}，组里的数据情况{group.shape}:")
    print(group.iloc[:,0:3])
    print("-------------------")
```

## 4)按多字段分组

```python
salary_mean = df.groupby(['department_id','job_id'])[['salary','commission_pct']].mean()
salary_mean.info()      # 按多个字段分组后得到的索引为复合索引
salary_mean.reset_index()

# 也可以在分组的时候通过as_index = False参数（默认是True）重置索引
salary_mean1 = df.groupby(["department_id", "job_id"], as_index=False)[["salary", "commission_pct"]].mean()
salary_mean1
```
