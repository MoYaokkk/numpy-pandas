# Series使用apply

```python
import pandas as pd

def func(item):
    if item <= 2:
        return item * 20
    else:
        return item * 10

s = pd.Series([1, 2, 3, 4, 5])

s.apply(func)

s + 10

s.apply(lambda x: x + 10)



# apply()方法会将自己没有匹配上的参数，在调用func的时候作为func的参数传递过去,必须通过关键字传参
def func1(item,p1):
    return item * p1
s.apply(func1,p1=3)
```

# DataFrame使用apply()

```python
def func(item):
    print(type(item))
    res = item.sum()
    return 100 if res < 100 else res


df = pd.DataFrame({"a": [10, 20, 30], "b": [40, 50, 60]})
# df.sum()
df.apply(func)

```

```python
# 注意：df.apply 一次只能处理一个 Series（当 axis=0 时处理列，当 axis=1 时处理行），但如下的函数 func中，获取同一个Series对象的两列数据a和b，所以不能直接使用 df.apply(f)按列处理了，需要指定axis=1,按行处理，才能获取一行中的不同列。

def func(s):
    return s["a"] / s["b"]

df = pd.DataFrame({"a": [10, 20, 30], "b": [40, 50, 60]})
df.apply(func, axis=1)
```

# 向量化函数

```python
import numpy as np
# 处理函数
@np.vectorize
def f(x, y):
    if y == 0:
        return np.nan
    return x / y

df = pd.DataFrame({"a": [10, 20, 30], "b": [40, 0, 60]})    # 字典

# f_vec = np.vectorize(f)     # 向量化,不然上面函数里是标量运算
# f_vec(df['a'],df['b'])
f(df['a'],df['b'])     # 在函数里进行向量化说明
```
