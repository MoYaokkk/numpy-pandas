# 透视表

```python
import pandas as pd
df = pd.read_csv(r"D:\AAAjobbing\actfirst\Numpy&Pandas\data\employees.csv")
# 统计每一个部门的平均薪水
df.groupby('department_id').agg({'salary': 'mean'})

# 透视表
df.pivot_table(values='salary',aggfunc='mean',index='department_id')
df.pivot_table(values='salary',aggfunc='mean',columns='department_id')
```

# 睡眠质量分析透视表

```python
df = pd.read_csv(r"D:\AAAjobbing\actfirst\Numpy&Pandas\data\sleep.csv")
df
```

```python
# 1）统计不同睡眠时间，不同压力等级下的睡眠质量
# 因为数据过于细致,所以对睡眠时间和压力等级进行cut分割
sleep_duration_stage = pd.cut(df['sleep_duration'],[0,5,6,7,8,9,10,11,12])
sleep_duration_stage
stress_level_stage = pd.cut(df['stress_level'],4)
stress_level_stage

df.pivot_table(values='sleep_quality',index=[sleep_duration_stage,stress_level_stage])      # 直接把对象拿过来,之前都是拿列名.但是总计margins可能会出错
```

```python
# 2）添加职业作为列维度
df.pivot_table(values='sleep_quality', index=[sleep_duration_stage, stress_level_stage],columns='occupation')
# 2）添加职业作为列维度
```

```python
# 3）添加性别作为第二个列维度]
df.pivot_table(values='sleep_quality', index=[sleep_duration_stage, stress_level_stage],columns=['occupation','gender'])
```
