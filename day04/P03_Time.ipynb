# Python中自带的日期类型datetime

```python
from datetime import datetime

date1 = datetime(year=2000, month=1, day=1)
date2 = datetime.now()
print(date1)  # 2000-01-01 00:00:00
print(date2)
print(date1.year)  # 2000
print(date1.month)  # 1
print(date1.day)  # 1
print(date2.weekday())      # 0~6星期一~星期日
print(date2.strftime("%A"))
print(date2 - date1)
```

# 提取TimeStamp的各个部分

```python
d = pd.Timestamp("2015-01-01 09:08:07.123456")
print(d.year)  # 2015
print(d.month)  # 1
print(d.day)  # 1
print(d.hour)  # 9
print(d.minute)  # 8
print(d.second)  # 7
print(d.microsecond)  # 123456
```

```python
# Series通过时间属性访问器对象访问
df = pd.read_csv(r"D:\AAAjobbing\actfirst\Numpy&Pandas\data\weather.csv")
df_date = pd.to_datetime(df["date"])
df['year'] = df_date.dt.year
df['month'] = df_date.dt.month
df['day'] = df_date.dt.day
df
```

# Period

```python
df["quarter"] = pd.to_datetime(df["date"]).dt.to_period("Q")  # 将 年-月-日 转换为 年季度
df.set_index('quarter', inplace=True)
df.info()
```
